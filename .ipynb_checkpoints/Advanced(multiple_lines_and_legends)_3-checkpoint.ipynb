```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4]
y1=[10,20,30,40]
y2=[20,30,40,50]
```

```python
plt.plot(x,y1,label='2025 sales',marker='o',color='b')#plotting y1
plt.plot(x,y2,label='2026 sales',marker='o',color='r')#plotting y2
plt.title('sales')
plt.xlabel('month')
plt.ylabel('sales')
plt.legend()#it is for showing legend (index)
plt.show()
```
