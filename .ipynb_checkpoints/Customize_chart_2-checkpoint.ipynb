```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4]
y=[10,20,30,40]
```

```python
plt.figure(figsize=(4,4))#we can change or adjust chart size with the help of plt.figure
plt.plot(x, y, color='b', marker='o', linestyle='-', linewidth=3, markersize=12)
plt.title('line chart')#it is for chart title
plt.xlabel('x-label')
plt.ylabel('y-label')
plt.show()
```
