```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    'Month':['jan','feb','mar','apr'],
    'salary':[20000,30000,35000,40000]
              }
```

```python
df=pd.DataFrame(data)
df
```

```python
plt.figure(  figsize=(3,4))
plt.bar(df['Month'],df['salary'],color='g')
plt.title('bar graph')
plt.xlabel('month')
plt.ylabel('salary')
plt.show()
```
