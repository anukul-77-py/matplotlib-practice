```python
#pie chart is used to show part-to-whole relationships/data
```

```python
import matplotlib.pyplot as plt
```

```python
categories=['A','B','C','D']
sales=[12,15,39,45]
```

```python
plt.pie(sales,labels=categories,  autopct='%1.1f%%',    rotatelabels=90)
plt.title('pie chart example ')
plt.show()
```
