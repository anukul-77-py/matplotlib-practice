```python
import matplotlib.pyplot as plt
```

```python
#data1
x=['A','B','C','D']
x1=[10,20,30,40]

```

```python
#data 2
categories=['A','B','C','D']
sales=[3000,4000,6500,8000]
```

```python
#figure 1 bar graph
plt.figure(  figsize=(10,6))
plt.subplot(1,2,1)#row , columns , position
plt.bar(x,x1,)
plt.title("bar graph")
plt.xlabel ('x-label')
plt.ylabel('y-label')

#figure 2 pie chart
plt.subplot(1,2,2)
plt.pie(sales,labels=categories ,autopct='%1.1f%%')
plt.title("pie chart")

plt.legend()        
plt.show()
```
